```python
import pandas as pd

netflix = pd.read_csv("netflix_titles.csv")

netflix.head()
```

```python
print("Shape:", netflix.shape)
print("\nColumn names:", netflix.columns.tolist())
print("\nMissing values per column:\n", netflix.isnull().sum())
print("\nDuplicate rows:", netflix.duplicated().sum())
```

```python
# Chhoti missing values wali rows hatao (bohot kam hain, farq nahi padega)
netflix = netflix.dropna(subset=['date_added', 'rating', 'duration'])

# Badi missing values wale columns ko "Unknown" se bharo
netflix['director'] = netflix['director'].fillna('Unknown')
netflix['cast'] = netflix['cast'].fillna('Unknown')
netflix['country'] = netflix['country'].fillna('Unknown')

# Confirm karo sab theek ho gaya
print("Missing values ab:\n", netflix.isnull().sum())
```

```python
# Check 1: kya 'date_added' asal mein date hai, ya sirf text?
print("date_added ka type:", netflix['date_added'].dtype)

# Check 2: 'type' column mein sirf 'Movie' aur 'TV Show' hone chahiye
print("\ntype ki unique values:\n", netflix['type'].value_counts())

# Check 3: 'rating' mein koi ajeeb/galat category to nahi
print("\nrating ki unique values:\n", netflix['rating'].value_counts())

# Check 4: kya koi TITLE repeat ho raha hai (chahe poori row match na ho)
print("\nRepeat hone wale titles:", netflix['title'].duplicated().sum())
```

```python
netflix['date_added'] = pd.to_datetime(netflix['date_added'].str.strip())

print("Naya type:", netflix['date_added'].dtype)
netflix['date_added'].head()
```

```python
netflix.to_csv("netflix_clean.csv", index=False)
```
